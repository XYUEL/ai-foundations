```python
#### 1. Import the numpy package under the name `np` (★☆☆)
import numpy as np

#### 3. Create a null vector of size 10 (★☆☆)
vec3 = np.zeros(10)

#### 4. How to find the memory size of any array (★☆☆)
print(vec3.nbytes)

#### 6. Create a null vector of size 10 but the fifth value which is 1 (★☆☆)
vec6 = np.empty(10)
vec6[4] = 1
print(vec6)

#### 7. Create a vector with values ranging from 10 to 49 (★☆☆)
vec7 = np.arange(10,50)

#### 8. Reverse a vector (first element becomes last) (★☆☆)
vec8 = np.arange(10)
print(vec8)
vec8 = vec8[::-1]
print(vec8)

#### 9. Create a 3x3 matrix with values ranging from 0 to 8 (★☆☆)
matrix9 = np.arange(9).reshape(3,3)
print(matrix9)

#### 10. Find indices of non-zero elements from [1,2,0,0,4,0] (★☆☆)
print(np.nonzero([1,2,0,0,4,0]))

```

```python
#### 11. Create a 3x3 identity matrix (★☆☆)
matrix11 = np.eye(3)
print(matrix11)

#### 13. Create a 10x10 array with random values and find the minimum and maximum values (★☆☆)
matrix13 = np.random.random((10,10))
print(matrix13)
print(matrix13.max())
print(matrix13.min())

```

```python
#### 14. Create a random vector of size 30 and find the mean value (★☆☆)
vec14 = np.random.random(30)
print(vec14)
print(vec14.mean())


#### 15. Create a 2d array with 1 on the border and 0 inside (★☆☆)
matrix15 = np.ones((3,4))
matrix15[1:-1,1:-1] = 0
print(matrix15)

```

```python
#### 19. Create a 8x8 matrix and fill it with a checkerboard pattern (★☆☆)
matrix19 = np.ones((8,8))
matrix19[::2,::2] = 0
matrix19[1::2,1::2] = 0
print(matrix19)

#### 22. Normalize a 5x5 random matrix (★☆☆)
matrix22 = np.random.random((5,5))
print(matrix22)
mean22 = matrix22.mean()
std22 = matrix22.std()
matrix22 = (matrix22 - mean22)/std22
print(matrix22)

#### 24. Multiply a 5x3 matrix by a 3x2 matrix (real matrix product) (★☆☆)
matrix2401 = np.ones((5,3))
matrix2402 = np.ones((3,2))
print(matrix2401 @ matrix2402)


```
